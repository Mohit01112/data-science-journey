## wordnetLemmetization


```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going",pos='v')
```

```python
## CLASSIFICTAION
## here all the words are adjectives
words = ["running", "runs", "easily", "fairly", "arguing", "connection", "connects"]
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='a'))
```

```python
## CLASSIFICTAION
## here all the words are verbs bcz pos='v'
words = ["running", "runs", "easily", "fairly", "arguing", "connection", "connects"]
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```

```python
## CLASSIFICTAION
## here all the words are nouns bcz pos='n'
words = ["running", "runs", "easily", "fairly", "arguing", "connection", "connects"]
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='n'))
```
