```python
## CLASSIFICTAION
words = ["running", "runs", "easily", "fairly", "arguing", "connection", "connects"]
```

```python
## porter stemming
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"----->"+stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

### REGEXPSTEMMER

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stem=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stem.stem("eating")
```

```python
reg_stem.stem("ingeating")
```

```python
 from nltk import SnowballStemmer
sbstemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"------------->"+sbstemmer.stem(word))
```
