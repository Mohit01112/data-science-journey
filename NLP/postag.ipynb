```python
para=""" My dear young friends,

Whenever I interact with the youth of our nation, I see a spark. I see an ignition of dreams that has the power to transform the world. Today, let us look at our journey through a unique lens—the lens of technology and human nature. 

In the field of computer science, when we train a computer to read human language, we use a process called filtering 'stop words'. Stop words are the most common words in a language—words like 'is', 'the', 'and', 'but', or 'of'. They fill the pages, they occupy space, and they create noise. But when a computer wants to find the true meaning, the deep core knowledge of a book, it must systematically remove these stop words. Only when the noise is eliminated do the powerful words—words like 'Dream', 'Action', 'Perseverance', and 'Success'—shine through.

My friends, look closely at your own lives today. We, too, are constantly surrounded by personal 'stop words'. 

What are the stop words of the human mind? 
They are 'Doubt'. They are 'Fear'. They are the small, repetitive noises of 'I cannot', 'Tomorrow', 'What if I fail?', and 'But'. These words fill your daily thoughts. They occupy your mental bandwidth. They consume your precious time, yet they add absolutely zero value to your ultimate destiny. 

If you want to achieve your dreams, you must become the master programmer of your own mind. You must run a pipeline that filters out these negative stop words. 

When you eliminate the noise of doubt, what remains? What remains is your core purpose. What remains is your untamed potential. 

Remember my words: 'Great dreams of great dreamers are always transcended.' But to transcend, you must clear the clutter. Do not let the common, noisy distractions of the world drown out the unique, powerful message your life is meant to deliver. 

Cleanse your thoughts. Remove the stop words of hesitation. Ignite your minds, act with courage, and fly toward your destiny. 

Thank you, and Jai Hind.
"""
```

```python
import nltk
sentences=nltk.sent_tokenize(para)
```

```python
from nltk.corpus import stopwords
sentences=nltk.sent_tokenize(para)
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[word for word in words if word not in set(stopwords.words('english'))]
    ## sentences[i]=" ".join(words)## converting all the words into sentences
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)
```
