```python
## create columns  trandsformers with 3 types of transformer
num_features=x.select_dtypes(exclude='str').columns
onehot_columns=['seller_type','fuel_type','transmission_type']
label_encoder_columns=['model']
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
numeric_transformer=StandardScaler()
oh_transformer=OneHotEncoder(drop='first')

preprocessor=ColumnTransformer(
    [
        ('OneHotEncoder',oh_transformer,onehot_columns),
        ('StandardScaler',numeric_transformer,num_features)
    ],remainder='passthrough'
)
```

```python
x=preprocessor.fit_transform(x)
```

```python
x_train ,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_model(true, predicted):
    mae = mean_absolute_error(true, predicted)  # 💡 Added so 'mae' exists
    mse = mean_squared_error(true, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(true, predicted)              # 💡 Renamed to 'r2' to stop the crash
    
    return mae, rmse, r2 
```

```python
models={
    "Linear regression":LinearRegression(),
    "Lasso":Lasso(),
    "Ridge":Ridge(),
    "K-neighbours regressor":KNeighborsRegressor(),
    "Decision Tree": DecisionTreeRegressor(),
    "Rando forest regressor":RandomForestRegressor()
}
```

```python
for i in range (len(list(models))):
    model=list(models.values())[i]
    model.fit(x_train,y_train)

    ##v make prediction
    y_train_pred=model.predict(x_train)
    y_test_pred=model.predict(x_test)

    ## evaluacte train and test
    model_train_mae,model_train_rmse,model_train_r2=evaluate_model(y_train,y_train_pred)
    model_test_mae,model_test_rmse,model_test_r2=evaluate_model(y_test,y_test_pred)

    print(list(models.keys())[i])
    print("model performance for training set")
    print("root mean squared error :{:.4f}".format(model_train_rmse))
   
    print(" mean absolute error :{:.4f}".format(model_train_mae))
    print("r2 score:{:.4f}".format(model_train_r2))

    print('---------------------------------------------')
    
    print("model performance for test set")
    print("root mean squared error :{:.4f}".format(model_test_rmse))
   
    print(" mean absolute error :{:.4f}".format(model_test_mae))
    print("r2 score:{:.4f}".format(model_test_r2))
    print("="*35)
    print("\n")
    
```

```python
## hyper paramter tunning
knn_params={"n_neighbours":[2,3,10,20,40,50]}
rf_params={"max_depth":[5,8,None,10],
          "max_features":[5,7,"auto",8],
          "min_samples_split":[2,8,15,20],
          "n_estimators":[100,200,500,1000]}
```

```python
randomcv_models=[('KNN',KNeighborsRegressor(),knn_params),
                ("RF",RandomForestRegressor(),rf_params)]
```

```python
from sklearn.model_selection import RandomizedSearchCV
model_param={}
for name, model, params in randomcv_models:
    
    # 💡 Safety Check: Make sure n_iter isn't larger than total possible combinations
    # Optional but highly recommended if your search space is tight
    
    random = RandomizedSearchCV(
        estimator=model,
        param_distributions=params,
        n_iter=100,          # 💡 Reduced from 100 to 20 for faster execution; increase later if needed
        cv=3,
        verbose=2,
        n_jobs=-1,          # Uses all CPU cores    # 💡 Added for reproducible, consistent results
    )
    
    random.fit(x_train, y_train)
    model_param[name] = random.best_params_

# Corrected printing loop
for model_name in model_param:
    print(f"\n----------------- Best params for {model_name} -----------------")
    print(model_param[model_name])
```

```python
models={
    "K-neighbours regressor":KNeighborsRegressor(n_neighbors=10),
    "Rando forest regressor":RandomForestRegressor(n_estimators=500, min_samples_split=2,max_features=5,max_depth=None)
}
for i in range (len(list(models))):
    model=list(models.values())[i]
    model.fit(x_train,y_train)

    ##v make prediction
    y_train_pred=model.predict(x_train)
    y_test_pred=model.predict(x_test)

    ## evaluacte train and test
    model_train_mae,model_train_rmse,model_train_r2=evaluate_model(y_train,y_train_pred)
    model_test_mae,model_test_rmse,model_test_r2=evaluate_model(y_test,y_test_pred)

    print(list(models.keys())[i])
    print("model performance for training set")
    print("root mean squared error :{:.4f}".format(model_train_rmse))
   
    print(" mean absolute error :{:.4f}".format(model_train_mae))
    print("r2 score:{:.4f}".format(model_train_r2))

    print('---------------------------------------------')
    
    print("model performance for test set")
    print("root mean squared error :{:.4f}".format(model_test_rmse))
   
    print(" mean absolute error :{:.4f}".format(model_test_mae))
    print("r2 score:{:.4f}".format(model_test_r2))
    print("="*35)
    print("\n")
    
```
